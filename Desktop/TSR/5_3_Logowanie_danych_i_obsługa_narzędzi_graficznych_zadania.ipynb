# Rejestrowanie rosbaga
Należy wykorzystać polecenia z terminala.

### Zadanie 1
Zarejestrować rosbaga dla topiców od położenia robotów. Nagrywanie przerwać po 10s.

### Zadanie 2
Zarejestrować rosbaga o nazwie robot_poses.bag rejestrującego położenie robotów. Zakończenie rejestrowania odbywa się ręcznie przez użytkownika przy pomocy ctrl+c.

### Zadanie 3
Zarejestrować rosbaga z prędkościami robotów trwającego 5s. Należy zapisywać tylko 3 ostatnie rosbagi. Pozostałe mają się kasować.

### Zadanie 4
Zarejestrować dla każdego topic'a od prędkości robotów tylko 20 wiadomości.

### Zadanie 5
Zarejestrować wszystkie topic'i dla node'a turtlesim.

# Odtwarzanie rosbaga

### Zadanie 6
Odtworzyć z rosbaga TSR.bag dane dla topic'u /robot1/cmd_vel.

### Zadanie 7
Odtworzyć rosbaga z dwukrotnie większą prędkością.

### Zadanie 8
Rozpocząć odtwarzanie rosbaga od 5 sekundy.

## Set

```python
"""
set - jest to kolekcja wielu elementów.
Elementy są:
 - nieposortowane,
 - nieedytowalne (ale można dodawać i usuwać elementy),
 - nieindexowane.
Zbiór nie zawiera zduplikowanych wartości.
Zapisywane są w sześciennych nawiasach.
""" 
punkty = {"A", "C", "D"}

# mogą przechowywać elementy różnego typu
dane = {"A", "C", 3, "D", True}
```

```python
# dodanie nowego elementu
dane.add(34)
```

```python
"""
Niektóre metody:
 - clear() - wyczyszczenie listy
 - copy() - zwraca kopie aktualnej listy
 - pop() - usuwa losowy element ze zbioru
 - remove(element) - usuwa przekazany element jako argument ze zbioru
 - difference(set2) - zwraca różnicę elementów pomiędzy pierwszym
                      a drugim zbiorem set2
 - update(set2) - aktualizuje zbiór poprzez dodanie wartści z 
                  drugiego zbioru set2
"""
```

```python
# Aktualizacja zbioru - update
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
zbior_1.update(zbior_2)
print(zbior_1)
print("modyfikacja zbioru 2")
zbior_2.remove("e")
print(zbior_1)
print("Modyfikacja zbior_2 po aktualizacji zbior_1 nie wpływa na zbior_1")
```

```python
# Różnica zbiorów - difference
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
print("zbior_1 - zbior_2: " + str(zbior_1.difference(zbior_2)))
print("zbior_2 - zbior_1: " + str(zbior_2.difference(zbior_1)))
```

```python
# Zatrzymanie elementów w zbiorze, które występują w innym zbiorze, usunięcie pozostałych
# Modyfikacja zbior_1
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
zbior_1.intersection_update(zbior_2)
print(zbior_1)
```

```python
# Utworzenie nowego zbioru z elementami wspólnymi obu zbiorów
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
nowy_zbior = zbior_1.intersection(zbior_2)
print("zbior_1=" + str(zbior_1))
print("zbior_2=" + str(zbior_2))
print("nowy_zbior=" + str(nowy_zbior))
```

```python
# Utworzenie nowego zbioru z elementami wspólnymi obu zbiorów
# Wartości True i 1 są sobie równoważne.
# W nowym zbiorze pojawia się 1 jako część wspólna.
zbior_1 = {"a", "b", "c", "d", "e", True}
zbior_2 = {"b", "e", "g", "h", 1}
nowy_zbior = zbior_1.intersection(zbior_2)
print(nowy_zbior)
```

```python
# Zatrzymanie elementów różniących oba zbiory.
# Elementy wspólne są pomijane.
# Aktualizacja zbior_1
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
zbior_1.symmetric_difference_update(zbior_2)
print(zbior_1)
```

```python
# Utworzenie nowego zbioru z elementów, które nie powtarzają się w obu zbiorach.
zbior_1 = {"a", "b", "c", "d", "e"}
zbior_2 = {"b", "e", "g", "h"}
nowy_zbior = zbior_1.symmetric_difference(zbior_2)
print("zbior_1=" + str(zbior_1))
print("zbior_2=" + str(zbior_2))
print("nowy_zbior=" + str(nowy_zbior))
```

```python
# Usuwanie elementu ze zbioru - remove
zbior = {"a", "b", "c", "d", "e"}
zbior.remove("c")
print(zbior)
```

```python
# Usuwanie elementu ze zbioru - remove - błąd - element nie istnieje
zbior = {"a", "b", "c", "d", "e"}
zbior.remove("z")
```

```python
# Usuwanie elementu ze zbioru - discard
zbior = {"a", "b", "c", "d", "e"}
zbior.discard("c")
print(zbior)
```

```python
# Usuwanie elementu ze zbioru - discard - brak błędu - element nie istnieje
zbior = {"a", "b", "c", "d", "e"}
zbior.discard("z")
print(zbior)
```

# Operacje na obiektach

```python
"""
Operacje matematyczne.
 - dodawania(+), 
 - odejmowania(-), 
 - mnożenia(*), 
 - dzielenia(/), 
 - operacji modulo(%),
 - potęgowania (**),
 - dzielenia bez reszty (//).

"""

# Podnoszenie do potęgi
kwadrat = 3**2
szescian = 6**3
print("kwadrat " + str(kwadrat))
print("sześcian " + str(szescian))

# dzielenie bez reszty
wynik = 15//2
print("wynik dzielenia 15/2 bez reszty: " + str(wynik))
```

```python
"""
Operatory porównania.
 - równe (==),
 - różne (!=),
 - większe niż (>),
 - mniejsze niż (<),
 - większe lub równe (>=),
 - mniejsze lub równe (<=),
"""
a = 4
b = 7
print("a == 4  " + str(a == 4)) # True
print("a != 4  " + str(a != 4)) # False
print("a == 5  " + str(a == 5)) # False
print("a > 1   " + str(a > 1))  # True
#zmiana wartości wyrażenia na przeciwną
print("not a == 4  "+ str(not a == 4))
```

```python
"""
Operatory logiczne.
 - and - iloczyn logiczny
 - or - suma logiczna
 - not - zaprzeczenie
"""
print("a == 4 and b > 3   " + str(a == 4 and b > 3)) #True
print("a == 2 or b > 2    " + str(a == 2 or b > 2))  #True
```

```python
# Do sprawdzenia czy podany operator znajduje się w zbiorze służy operator in
tablica = ["abc", "def", "ghi"]
print("def" in tablica)  # true
print("test" in tablica) # false
```

```python
# Do sprawdzenia czy zmienne wskazują na ten sam obszar w pamięci służy operator is
x = [5, 3, 3]
y = [2, 6, 3]

tablica2 = [4, x, y]

print("x == tablica2[1]  " + str(x == tablica2[1]))  # True
print("x is tablica2[1]  " + str(x is tablica2[1]))  # True
print("x is tablica2     " + str(x is tablica2))     # False
```

# Zmienne globalne
Znajdują się na zewnątrz funkcji lub klasy. Do ich modyfikacji w funkcji lub klasie wykorzystuje się
słowo kluczowe **global**.

```python
x = 10

def funkcja():
    global x # wskazuje, że w funkcji będzie modyfikowana zmienna globalna x
    x = 20 

funkcja()

print(x)
```

# Klasy

```python
class Robot:
    def __init__(self, name, x, y):
        # inicjalizacja klasy, wykonuje się zawsze, gdy tworzony jest nowy obiekt
        # self. - wskazuje na zmienne należące do danej instancji klasy
        self.name = name
        self.x = x
        self.y = y
        
    def print_pose_info(self): # jedna z metod "print_pose_info" klasy "Robot"
        print("Pose x={:f}, y={:f}".format(self.x, self.y))
    
    def move_right(self):
        self.x += 1
        
    def move_left(self):
        self.x -= 1
        
    def move_up(self):
        self.y += 1
        
    def move_down(self):
        self.y -= 1
        
    def get_distance_to_start(self):
        # metoda zwraca odległość robota od startu (0,0)
        return math.sqrt(self.x*self.x + self.y+self.y)
```

```python
# Inicjalizacja klas Robot
robot1 = Robot("r1", 0, 0)
robot2 = Robot("r2", 5, 2)
```

```python
# Odczyt wartości atrybutu klasy
robot2.name
```

```python
# Użycie metody klasy
robot2.move_right()
robot2.print_pose_info()
```

```python
# Użycie metody zwracającej jakąś wartość, która została zapisana w zmiennej 
# dist_to_start
dist_to_start= robot2.get_distance_to_start()
print("Distance to start: ", dist_to_start)
```

## Klasa z atrybutem, który jest inną klasą

```python
class Wheel:
    def __init__(self, diameter, width):
        self.diameter = diameter
        self.width = width
        
class Robot:
    def __init__(self, name, x, y):
        # inicjalizacja klasy, wykonuje się zawsze, gdy tworzony jest nowy obiekt
        # self. - wskazuje na zmienne należące do danej instancji klasy
        self.name = name
        self.x = x
        self.y = y
        self.left_wheel = Wheel(2,1)
        self.right_wheel = Wheel(2,1)
    
    def print_wheels_diameter(self):
        left_wheel = str(self.left_wheel.diameter)
        right_wheel = str(self.right_wheel.diameter)
        print("Left wheel: " + left_wheel + " right wheel: " + right_wheel)
```

```python
robot1 = Robot("r1", 0, 0)
# Odczyt średnicy koła lewego
print(robot1.left_wheel.diameter)
# Odczyt średnicy koła prawego
print(robot1.right_wheel.diameter)
```

```python
# utworzenie obiektu koła i przypisanie go do atrybutów z left i right wheel
new_wheel = Wheel(3,4)
robot1 = Robot("r1", 0, 0)
robot1.left_wheel = new_wheel
robot1.right_wheel = new_wheel
robot1.print_wheels_diameter()
```

```python
# Modyfikacja średnicy obiektu koła powoduje modyfikacje kół w robocie
new_wheel.diameter = 2
robot1.print_wheels_diameter()
```

```python
# Modyfikacja średnicy obiektu koła w atrybucie robota również powoduje zmianę
robot1.left_wheel.diameter = 2.4
robot1.print_wheels_diameter()
print("średnica w obiekcie koła: " + str(new_wheel.diameter))
```

# Klasa z Publisherem i Subscriberem - ROS - przykład

```python
import rospy
from geometry_msgs.msg import Twist, Pose
rospy.init_node('test')
# Dla danej klasy w projekcie najlepiej będzie utworzyć nowego node'a 
# odpowiedzialnego za obsługę interfejsu czy robota.
```

```python
class Robot:
    def __init__(self, name, x, y):
        self.name = name
        # Dodanie atrybutów z 
        self.velocity_pub = rospy.Publisher('/cmd_vel', Twist, queue_size=10)
        self.pose_sub = rospy.Subscriber('/pose_topic',Pose, self.callback_pose)
        self.pose = Pose()
        
    def callback_pose(self, data):
        """
        Metoda wywołuje się gdy na topicu /pose_topic pojawi się dowolna wiadomość.
        Przerwany zostaje aktualnie wykonywany program i aktualizowana jest wartość
        atrybutu pose. W pętli sterującej wartość ta może być wykorzystywana do 
        sterowania robotem.
        """
        self.pose = data
        
    def run(self):
        while True:
            msg = Twist()
            self.velocity_pub.publish(msg)
```

```python
# wysłanie raz prędkości sterującej
robot1 = Robot("r1", 0, 0)
robot1.run()
```

```python
# wysyłnaie prędkości sterujących w pętli dopóki roscore jest aktywny
robot1 = Robot("r1", 0, 0)
r = rospy.Rate(10) # 10hz

while not rospy.is_shutdown():
    # pętla odpowiedzialna za wysyłanie prędkości sterujących robotem. Działa
    # z częstotliwością 10Hz
    robot1.run() 
    r.sleep()
```

# Klasa z GUI

```python
import ipywidgets as widgets

class GUI:
    def __init__(self):
        # Atrybuty z kontrolkami: 2 kontrolki do wprowadzania tekstu i kontrolka 
        # z przyciskiem
        self.text_a = widgets.Text(value='test', disabled=False, 
                                   description="Text A:")
        self.text_b = widgets.Text(value='wersja A', disabled=False, 
                                   description="Text B:")
        self.print_button = widgets.Button(
            value=False,
            description='print text',
            disabled=False,
            button_style='', # 'success', 'info', 'warning', 'danger' 
            # or ''
            tooltip='Description',
            icon='check' # (FontAwesome names without the `fa-` prefix)
        )
        # Dodanie obsługi kliknięcia przycisku. Po kliknięciu wywoła się 
        # metoda print_text z tej klasy (GUI)
        self.print_button.on_click(self.print_text)

    def display_gui(self):
        # metoda odpowiedzialna za wyświetlenie interfejsu użytkownika
        # na tym poziomie można ustawic opis z tworzeniem siatki z przyciskami
        display(self.text_a)
        display(self.text_b)
        display(self.print_button)
        
    def print_text(self, button_data):
        # metoda wywołująca się po kliknięciu przycisku z napisem print_text
        print(self.text_a.value + self.text_b.value)
```

```python
# Utworenie obiektu GUI
gui = GUI()
# Wyświetlenie interfejsu
gui.display_gui()
```

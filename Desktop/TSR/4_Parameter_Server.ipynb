# ROS Parameter Server

Parameter Server jest zbiorem wartości parametrów odczytywanych przez nody w ROS. Parametry mogą przechowywać informacje o konfiguracji sprzętowej lub innych charakterystycznych parametrach. Serwer parametrów wykorzystuje 1 do 1 odpowiednik formatu YAML.
## yaml przykład

Istnieje możliwość zmiany pojedynczych parametrów, ale można je też nadpisać słownikiem w rosparam. Oprócz samych wartości można wpisywać inne matematyczne operacje.

```python
# Komendy dla polecenia rosparam
!rosparam --help
```

### lista dostępnych parametrów

```python
!rosparam list
```

### odczyt pojedynczego parametru

```python
# rosparam get cała_nazwa_parametru
!rosparam get /turtlesim/background_b
```

```python
# w przestrzeni nazw turtlesim znajdują się parametry:
# background_b, background_g, background_r
!rosparam get /turtlesim
```

### ustawianie nowego parametru

```python
!rosparam set robot_1/tryb_pracy "auto"
```

```python
# na liście parametrów pojawił się parametr /robot_1/tryb_pracy
!rosparam list
```

```python
# wartość tego parametru wynosi
!rosparam get /robot_1/tryb_pracy
```

```python
# wykorzystanie słownika przy ustawianiu parametru
!rosparam set robot_2 "{v_max: 4, omega_max: 2}"
```

```python
# na liście parametrów pojawił się parametry dla robot_2
!rosparam list
```

```python
!rosparam delete /robot_1/tryb_pracy
```

# Obsługa w języku Python

## Lista dostępnych parametrów

```python
import rospy
rospy.get_param_names()
```

## Odczyt wartości parametru

```python
print(rospy.get_param("turtlesim/background_r"))
print(rospy.get_param("turtlesim/background_g"))
print(rospy.get_param("turtlesim/background_b"))
```

Ustawienie nowych wartości parametrów tła symulacji turtlesim
Aby wprowadzone zmiany były widoczne należy wywołać dodatkowo serwis */clear*

```python
rospy.set_param("turtlesim/background_r", 255)
rospy.set_param("turtlesim/background_g", 0)
rospy.set_param("turtlesim/background_b", 0)
```

```python
# dla symulacji z turtlesim wywołanie serwisu clear po zmianie parametrów tła powoduje zmianę
# koloru tła
!rosservice call /clear
```

```python
# surowe obekty w Pythonie
rospy.set_param('robot_on', False)
# przekazanie wartości jako słownik
rospy.set_param('gains', {'p': 1, 'i': 2, 'd': 3})
```

```python
# biblioteka do obsługi serwera parametrów (alternatywne podejście zamiast rospy.set_param())
import rosparam
# yaml string jako wartość
rosparam.set_param('robot_on', "true")
rosparam.set_param('gains', "{'p': 1, 'i': 2, 'd': 3}")
```

```python
!rosparam get /robot_on
```

Ustawianie kilku różnych parametrów w określonej przestrzeni nazw. Na przykładzie kolorów RGB tła można by ustawić je w przestrzeni nazw **color** i zgrupować razem. Takie ustawienie parametrów nie będzie miało wpływu na środowisko symulacyjne (turtlesim), gdyż nie są one obsługiwane.

```python
grupa = "color"
rospy.set_param(grupa+"/r", 41)
rospy.set_param(grupa+"/g", 231)
rospy.set_param(grupa+"/b", 0)
```

```python
rospy.get_param("color")
```

## Usuwanie parametrów 
Do usunięcia pojedynczego parametru należy podać jego pełną nazwę.

```python
rospy.delete_param("/color/r")
```

Można usunąć więcej parametrów znajdującej się w określonej przestrzeni nazw.

```python
rospy.delete_param("/color/")
```

W trakcie próby usunięcia nieistniejącego parametru pojawia się błąd

```python
rospy.delete_param("/color/")
```

```python
# Pojawił się błąd KeyError, aby uniknąć nagłego przerwania działania programu można obsłużyć
# wyjątek z wykorzystaniem składni try except. Błąd typu KeyError, który pojawia się w try
# jest obsłużony w części except i pojawia się informacja "value not set". Nie pojawia się
# już błąd tak jak w powyższym przykładzie.
try:
    rospy.delete_param('to_delete')
except KeyError:
    print("value not set")
```

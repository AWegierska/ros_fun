# ROS Parameter Server

Parameter Server jest zbiorem wartości parametrów odczytywanych przez nody w ROS. Parametry mogą przechowywać informacje o konfiguracji sprzętowej lub innych charakterystycznych parametrach. Serwer parametrów wykorzystuje 1 do 1 odpowiednik formatu YAML.

Dostępne typu parametrów:
- 32-bit integers
- booleans
- strings
- doubles
- iso8601 dates
- lists
- base64-encoded binary data

## yaml przykłady plików konfiguracyjnych
### uproszczony
```
string: 'jakis tekst'
integer: 3948
float: 231.5 
boolean: false
list: [1.0, mixed list]
dictionary: {a: b, c: d}
```

  global_frame: map # tb3_0/odom

  static_map: false
  rolling_window: true

  update_frequency: 10.0
  publish_frequency: 10.0
  transform_tolerance: 0.5  

  static_map: false

Istnieje możliwość zmiany pojedynczych parametrów, ale można je też nadpisać słownikiem w rosparam. Dla kątów w reprezentacji stopnie/radiany możliwe jest zastosowanie wyrażeń pozwalających na konwersję wartości. Dopuszczalne jest użycie **pi**, które stanowi reprezentację liczby **PI**.

## Obsługa grupy parametrów
#### Dodanie parametrów do określonej przestrzeni nazw.

Ustawianie kilku różnych parametrów w określonej przestrzeni nazw. Na przykładzie kolorów RGB tła można by ustawić je w przestrzeni nazw **color** i zgrupować razem. Takie ustawienie parametrów nie będzie miało wpływu na środowisko symulacyjne (turtlesim), gdyż nie są one obsługiwane.

````python
grupa = "color"
rospy.set_param(grupa+"/r", 41)
rospy.set_param(grupa+"/g", 231)
rospy.set_param(grupa+"/b", 0)
````

#### Odczyt wartości parametrów z grupy color

````python
rospy.get_param("color")
````

Zapis parametrów do oddzielnych zmiennych ze składowymi kolorów.

````python
colors = rospy.get_param("color")
r, g, b = colors["r"], colors["g"], colors["b"]
print("Składowe kolorów r={}, g={}, b={}".format(r,g,b))
````

## Usuwanie parametrów

Do usunięcia pojedynczego parametru należy podać jego pełną nazwę.

````python
rospy.delete_param("/color/r")
````

Można usunąć więcej parametrów znajdującej się w określonej przestrzeni nazw.

````python
rospy.delete_param("/color/")
````

W trakcie próby usunięcia nieistniejącego parametru pojawia się błąd

````python
rospy.delete_param("/color/")
````

Pojawił się błąd KeyError, aby uniknąć nagłego przerwania działania programu można obsłużyć
wyjątek z wykorzystaniem składni **try except**. Błąd typu **KeyError**, który pojawia się w bloku **try** jest obsłużony w części **except** i pojawia się informacja "Parametr nie istnieje.". Nie pojawia się już błąd tak jak w powyższym przykładzie.

````python
try:
    rospy.delete_param('to_delete')
except KeyError:
    print("Parametr nie istnieje.")
    
````

### Dla parametru istniejącego na liście

````python
param_name = rospy.search_param('/robot_4/goal_tolerance/x')
v = rospy.get_param(param_name)

print("Nazwa parametru: {}".format(param_name))
print("Wartość parametru: {}".format(v))
````

### Dla nieistniejącego parametru

````python
param_name = rospy.search_param('/robot_4/data')
v = rospy.get_param(param_name) # Nie znaleziono parametru, wyświetlane wszystkie parametry

print("Nazwa parametru: {}".format(param_name))
print("Wartość parametru: {}".format(v))
````

## Dla częściowej nazwy parametru
Nie wiadomo w jakiej przestrzeni nazw znajduje się dany parametr. Uruchomienie fragmentu programu, który występował by w przestrzeni nazw node'a spowodowałoby, że wyszukiwanie zaczęłoby się od tej przestrzeni. Poniższy program nie jest wywołany w żadnej przestreni nazw dlatego tylko znaleziony zostaje główny parametr **background_b**.

````python
param_name = rospy.search_param('background_b')
v = rospy.get_param(param_name)

print("Nazwa parametru: {}".format(param_name))
print("Wartość parametru: {}".format(v))
````

Dla parametru **background_r** występuje on w różnych przestrzeniach nazw, ale poniższy kod nie jest wywołany pod żadną z nich, dlatego też nie zwraca informacji o znalezionej nazwie parametru.

````python
param_name = rospy.search_param('background_r')
v = rospy.get_param(param_name)

print("Nazwa parametru: {}".format(param_name))
print("Wartość parametru: {}".format(v))
````

Dla wyszukiwania parametru w nodzie, który jest uruchomiony w przestrzeni nazw **/foo/bar** przeszukiwanie rozpoczyna się od nazw w tej przestrzeni i stopniowo przechodzi do coraz wyższych warstw. Dla parametru **param_example** wyszukiwanie będzie odbywało się w następującej kolejności:
```
1. /foo/bar/param_example
2. /foo/param_example
3. /param_example
```

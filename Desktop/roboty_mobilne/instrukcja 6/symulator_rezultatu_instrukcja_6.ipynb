```python
#!/usr/bin/env python
import rospy
import time
from move_base_msgs.msg import MoveBaseActionResult

GOAL_REACHED = True
GOAL_FAIL = False
```

```python
def createResultMsg(goal_status):
    resultMsg = MoveBaseActionResult()
    if goal_status == True:
        resultMsg.status.status = 3
        resultMsg.status.text = "Goal reached"
    else:
        resultMsg.status.status = 4
        resultMsg.status.text = "Goal failure"    
    return resultMsg
```

```python
if __name__ == '__main__':
    rospy.init_node('listener',anonymous=True)
    pub_result = rospy.Publisher('/move_base/result', MoveBaseActionResult, queue_size = 1)
    time.sleep(1)
    pub_result.publish(createResultMsg(GOAL_REACHED))
```

```python
import rospy
import tf

if __name__ == '__main__':
    rospy.init_node('turtle_tf_listener')
    # tworzenie nowego obieku tf.TransformListener
    listener = tf.TransformListener()

    rate = rospy.Rate(1.0)
    while not rospy.is_shutdown():
        try:
            # listener.lookupTransform zwraca informację o transformacie pomiędzy /base_link
            # a /base_scan. Funkcja zwraca dwie listy:
            # - trans - związaną z przesunięciem liniowym /base_scan względem /base_link
            # - rot - związana z obrotem od /base_link do /base_scan
            (trans,rot) = listener.lookupTransform('/base_scan', '/base_link', rospy.Time(0))
        except (tf.LookupException, tf.ConnectivityException, tf.ExtrapolationException):
            continue
        print("trans=" + str(trans) + " rot=" + str(rot))
        rate.sleep()
```

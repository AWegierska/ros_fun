```python
import rospy
from turtlesim.msg import Color
```

```python
def callback(data):
    rospy.loginfo("r = %s, g = %s, b = %s", data.r, data.g, data.b)
    
def print_color():
    rospy.init_node('color_listener', anonymous=True)
    rospy.Subscriber("/turtle1/color_sensor", Color, callback)
    rospy.spin()
```

```python
if __name__ == '__main__':
    print_color()
```

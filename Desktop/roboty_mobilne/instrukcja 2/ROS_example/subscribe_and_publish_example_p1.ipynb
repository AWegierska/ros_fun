```python
#Przykladowy publisher dostarczajacy wiadomosci do drugiego skryptu, ktory jednoczesnie odbiera i publikuje nowa wiadomosc.
#import bibliotek i wiadomości rosowych
import rospy
from std_msgs.msg import Int32
rospy.init_node('old_pub', anonymous=True)
```

```python
def pub_old():
    pub = rospy.Publisher('old_int', Int32, queue_size=10)
    
    oldInt = Int32()
    oldInt.data = 13
    
    rate = rospy.Rate(1) # 1hz
    while not rospy.is_shutdown():
        pub.publish(oldInt)
        rate.sleep()
```

```python
if __name__ == '__main__':
    
    try:
        pub_old()
    except rospy.ROSInterruptException:
        pass
```

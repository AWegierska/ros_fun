```python
# Biblioteka umozliwiajaca tworzenie ROS Node i korzystanie z ROS
import rospy

# Import wiadomosci rosowej typu String z paczki std_msgs
# msg oznacza, ze importowane sa typy wiadomosci uzywane przez topic
# srv odnosiloby sie do typow wiadomosci uzywanych przez serwisy
from std_msgs.msg import String 
```

```python
def talker():
    # Deklaracja, ze node 'talker' bedzie publikowal wiadomosc 
    # na topicu 'chatter' typu String
    rospy.init_node('talker', anonymous=True)
    pub = rospy.Publisher('chatter', String, queue_size=10)
    
    rate = rospy.Rate(0.5) # 0.5hz czestotliwosc z jaka publikowane sa wiadomosci
    while not rospy.is_shutdown():
        hello_str = "hello world %s" % rospy.get_time()
        rospy.loginfo(hello_str) 
        pub.publish(hello_str) #publikowanie wiadomosci
        rate.sleep()
```

```python
if __name__ == '__main__':
    try:
        talker()
    except rospy.ROSInterruptException:
        pass
```

```python
import rospy
import tf
import csv
import numpy as np

rospy.init_node('path_saver', anonymous=True)

def setFileColumnName(fileName):
    data = ['x_pos', 'y_pos', 'z_pos', 'x_or', 'y_or', 'z_or' ,'w_or']
    with open(fileName+".csv", 'w') as csvFile:
        writer = csv.writer(csvFile)
        writer.writerow(data)
    csvFile.close()
    
def savePose(fileName,data):
    with open(fileName+".csv", 'a') as csvFile:
        writer = csv.writer(csvFile)
        writer.writerow(data)
    csvFile.close()
```

```python
if __name__ == '__main__':
    fileName = "circle"
    listener = tf.TransformListener()
    setFileColumnName(fileName)
    rate = rospy.Rate(5)
    while not rospy.is_shutdown():
        try:
            (trans,rot) = listener.lookupTransform("/odom", "/base_footprint", rospy.Time(0))
            angle = tf.transformations.euler_from_quaternion(rot)
            data = np.concatenate((trans, rot), axis=None)
            #rospy.loginfo("trans = %s", data)
            savePose(fileName,data)
        except (tf.LookupException, tf.ConnectivityException, tf.ExtrapolationException):
            continue
        rate.sleep() 
```

```python
import matplotlib.pyplot as plt
import itertools
import warnings
import pandas as pd
import numpy as np

fileName = "circle.csv"

pathData = pd.read_csv(fileName)
df = pd.DataFrame(columns = ['x','y'])
x = np.array(pathData.x_pos)
y = np.array(pathData.y_pos)

f = plt.figure(figsize=(10,10))
ax1 = f.add_subplot(1,1,1)
ax1.plot(x, y, 'black', linewidth=0.5)
ax1.set_xlabel('x [m]')
ax1.set_ylabel('y [m]')
ax1.set_title('Robot path')

max_left = min(x)
max_right = max(x)
print("srednica " + str(abs(max_right - max_left)))
plt.show()
```

```python
# Funkcja sinus. Argumentem wejściowym jest kąt w radianach.
alpha = math.radians(30) # konwersja stopni na radiany
np.sin(alpha)
```

```python
# Utworzenie wektora składającego się z 3 elementów
a = np.array([1,2,3])
print(a)
```

```python
# Utworzenie macierzy o rozmiarze 3x2
A = np.array([[1, 2],
              [3, 4],
              [5, 6]])
print(A)
print(f"Rozmiar macierzy: {A.shape}")
```

```python
# Utworzenie macierzy diagonalnej
Adiag = np.diag([1,2,3,4])
print(Adiag)
```

```python
# Transpozycja macierzy A
np.transpose(A)
```

```python
# Mnożenie macierzy
M1 = np.array([[1,0,0],
               [0,2,0],
               [0,0,3]])
M2 = np.array([[1,2,3],
               [3,4,5],
               [1,1,2]])
W = M1 @ M2  # operator mnożenia macierzy
print(W)
```

```python
# Wyznaczenie macierzy pseudoinwersji Moore’a–Penrose’a macierzy B
B = np.array([[1,0],
               [0,2],
               [0,5]])
Bpinv = np.linalg.pinv(B)
print(Bpinv)
```

```python
# Wyznaczanie macierzy odwrotnej do macierzy B
B = np.array([[1,0,0],
               [0,2,0],
               [0,0,3]])

Binv = np.linalg.inv(B)
print(Binv)
print("Sprawdzenie:")
print(B @ Binv)
```

```python
# konwersja danych z formatu numpy na listę. 
B = np.array([[1,0,0],
               [0,2,0],
               [0,0,3]])
B_lista = B.tolist()
print(f"Typ danych B: {type(B)}")
print(B)
print("\n")
print(f"Typ danych B_lista: {type(B_lista)}")
print(B_lista)
```

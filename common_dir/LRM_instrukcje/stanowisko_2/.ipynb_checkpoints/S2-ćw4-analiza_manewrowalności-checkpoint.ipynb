## 1. Stopnie swobody robota (Degrees of Freedom, DoF)

Stopnie swobody określają, jakie niezależne ruchy może wykonać robot w swojej przestrzeni operacyjnej.  
Dla robota mobilnego poruszającego się na płaszczyźnie 2D mogą istnieć trzy stopnie swobody:

- Przemieszczenie wzdłuż osi X,
- Przemieszczenie wzdłuż osi Y,
- Obrót wokół osi Z (zmiana orientacji).

Nie każdy robot posiada możliwość kontrolowania wszystkich trzech komponentów ruchu.


## 2. Mobilność robota (*Degree of Mobility, $\delta_m$*)

Mobilność robota określa, ile niezależnych stopni swobody jego ruchu można kontrolować **bezpośrednio** poprzez zmianę prędkości kół. Innymi słowy, informuje ona, w ilu kierunkach robot może się poruszać dzięki swoim napędom — bez dodatkowego sterowania kątem skrętu kół.

Mobilność definiuje się wzorem:

$$
\delta_m = 3 - \operatorname{rank}(\mathbf{C_1}(\beta_s))
$$

gdzie:
- $\mathbf{C_1}(\beta_s)$ – macierz ograniczeń bocznych wynikających z warunku braku poślizgu bocznego dla wszystkich kół.

**Interpretacja wartości $\delta_m$**
- $ \delta_m = 3 $ - pełna mobilność (robot może poruszać się w kierunku $x$, $y$ i obracać wokół własnej osi $Z$),
- $ \delta_m = 2 $ - np. robot różnicowy, który nie może poruszać się bokiem,
- $ \delta_m = 1 $ - np. trójkołowiec z jednym napędzanym kołem,
- $ \delta_m = 0 $ - robot statyczny (brak mobilności).

Rząd macierzy $\mathrm{rank}(\mathbf{C_1}(\beta_s))$ określa liczbę niezależnych ograniczeń (liczbę wektorów liniowo niezależnych w macierzy) nałożonych na ruch robota. Informuje, czy przekształcenie za pośrednictwem tej macierzy powoduje utratę wymiaru. Inaczej mówiąc, pokazuje, ile wymiarów ruchu zostaje zablokowanych przez warunki kinematyczne (np. brak poślizgu bocznego).

- Jeśli rząd macierzy $\mathbf{C_1}$ jest równy 3, to trójwymiarowy wektor prędkości zostaje przekształcony we wszystkich trzech wymiarach na inny wektor 3-elementowy. W tym przypadku wszystkie trzy stopnie swobody są ograniczone – robot nie może poruszać się w żadnym kierunku (brak mobilności).
- Jeśli rząd wynosi 2, oznacza to utratę rzędu macierzy co powoduje, że wektor 3-elementowy zostaje rzutowany na płaszczyznę. W tym przypadku jedno z możliwych przemieszczeń jest zablokowane – robot może się poruszać tylko w dwóch kierunkach (np. do przodu i obracać się).

Macierz $\mathbf{C_1}$ zawiera projekcje (rzuty) ruchów wszystkich kół na kierunki **prostopadłe do płaszczyzn ich toczenia** — czyli na kierunki, w których koła nie powinny się poruszać (poślizg boczny).  


<span style="color:red">**Wniosek: Im większy rząd macierzy $\mathbf{C_1}$, tym więcej ograniczeń kinematycznych ma robot, a jego mobilność jest mniejsza.**</span>

## 3. Sterowalność robota (Degree of Steerability, $ \delta_s $)

Sterowalność robota określa, ile niezależnych kół skrętnych może aktywnie zmieniać kierunek ruchu robota poprzez zmianę orientacji osi toczących się.  

Sterowalność definiuje się jako:

$$
\delta_s = \operatorname{rank}(\mathbf{C_{1s}}(\beta_s))
$$

gdzie:
- $\mathbf{C_{1s}}(\beta_s)$ – ograniczenia na brak poślizgu dla kół skrętnych z podstawionym znanym kątem skrętu $\beta_{s}$ dla danego koła

Interpretacja:

- Zmiana kąta koła skrętnego **nie wpływa bezpośrednio na pozycję robota** — efekt tej zmiany jest widoczny dopiero **podczas ruchu**, gdy koło zaczyna się toczyć w nowym kierunku.
- Wzrost rzędu macierzy $\mathbf{C_{1s}}$ oznacza, że robot posiada **więcej niezależnych możliwości sterowania kierunkiem ruchu** $\Rightarrow$ **większą manewrowalność**.
- Warto zauważyć, że **to samo koło skrętne** może jednocześnie **zmniejszyć mobilność** (ponieważ chwilowo dodaje ograniczenie na ruch), ale **zwiększyć sterowalność** (bo pozwala na zmianę orientacji koła).
    
    
### Typowe wartości:
- $ \delta_s = 0 $ → brak kół skrętnych (np. differential drive),
- $ \delta_s = 1 $ → np. trójkołowiec z jednym skrętnym kołem,
- $ \delta_s = 2 $ → np. robot z dwoma niezależnymi kołami skrętnymi.

## 4. Manewrowalność robota (*Degree of Maneuverability, $\delta_M$*)
Całkowita liczba stopni swobody, które można kontrolować (prędkością lub sterowaniem), to:

$$
\delta_M = \delta_m + \delta_s
$$

gdzie:
- $(\delta_m)$ – stopień mobilności robota,
- $(\delta_s)$ – stopień sterowalności robota,

> **Uwaga:** Dwa roboty o tej samej manewrowalności $ \delta_M $ **nie muszą być równoważne** pod względem zachowania.

Na przykład:
- Roboty z napędem różnicowym (differential drive) i roboty trójkołowe (tricycle) mają taką samą manewrowalność:  
  $ \delta_M = 2 $.

Ale różnice są istotne:
- Dla napędu różnicowego:  
  $ \delta_m = 2 $, $ \delta_s = 0 $ → **cała manewrowalność pochodzi z mobilności** (napędu).
  
- Dla trójkołowca:  
  $ \delta_m = 1 $, $ \delta_s = 1 $ → manewrowalność wynika częściowo z napędu, a częściowo z możliwości skrętu.

**Wniosek:** Mimo tej samej liczby dostępnych stopni swobody (manewrowalności), sposób ich realizacji może być zupełnie inny.

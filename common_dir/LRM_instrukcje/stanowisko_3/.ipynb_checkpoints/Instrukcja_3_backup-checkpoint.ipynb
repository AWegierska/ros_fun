```python
import rclpy
from rclpy.node import Node
from sensor_msgs.msg import JointState
from geometry_msgs.msg import Twist
from nav_msgs.msg import Odometry
import numpy as np
from geometry_msgs.msg import Quaternion
import math
```

- Wytłumaczenie jak działa nod i jaka jest architektura
- jak skonstruowany jest topic joint states i jak idą dane
- jak wygląda komunikacja między robotem a komputerem

```python
class Robot(Node):
    def __init__(self):
        super().__init__('robot_kinematics')
        
        self.x = 0
        self.y = 0
        self.theta = 0

        # Odbieranie wiadomości o prędkościach kół
        # z topica /joint_states
        self.subscription_joint_states = self.create_subscription(
            JointState,           # Message type
            '/joint_states_',    # Topic name
            self.joint_states_callback,  # Callback function
            10  # QoS profile depth
        )
        
        # Odbieranie wiadomości o zadanej prędkości robota
        # z topica /cmd_vel
        self.subscription_cmd_vel = self.create_subscription(
            Twist,           # Message type
            '/cmd_vel',       # Topic name
            self.cmd_vel_callback,  # Callback function
            10  # QoS profile depth
        )
        
        # Wysyłanie informacji o pozycji na topic /odometry
        self.publisher_odom = self.create_publisher(Odometry, '/odometry', 10)
        # Wysyłanie informacji o prędkościach na topic /odometry
        self.publisher_joint_states = self.create_publisher(JointState, '/cmd_joint_states', 10)
```

```python
x1 = 0.1871
y1 = 0.16325

alpha1 = np.atan2(y1,x1)
beta1 = np.atan2(x1,y1)
l1 = np.sqrt(x1*x1 + y1*y1)

x2 = 0
y2 = 0.144
alpha2 = np.atan2(y2,x2)
beta2 = np.atan2(x2,y2)
l2 = np.sqrt(x2*x2 + y2*y2)

x3 = 0
y3 = -0.144
alpha3 = np.atan2(y3,x3)
beta3 = np.atan2(x3,y3)
l3 = np.sqrt(x3*x3 + y3*y3)

x4 = 0.1871
y4 = -0.16325
alpha4 = np.atan2(y4,x4)
beta4 = np.atan2(x4,y4)
l4 = np.sqrt(x4*x4 + y4*y4)
```

obliczanie predkosci!!!

```python
def calculate_angle(x, y, vel_x, vel_y, vel_ang):
    R = math.sqrt(vel_x**2 + vel_y**2) / abs(vel_ang) #oblicz promień skrętu
    ICRx = -(vel_y/vel_ang)
    ICRy = (vel_x/vel_ang)
    
    l = math.sqrt(x**2 + y**2)
    
    c = math.sqrt((x - ICRx)**2 + (y - ICRy)**2)
    
    cos_gamma = (c**2 + R**2 - l**2) / (2 * c * R)
    gamma = math.acos(cos_gamma)  # Wynik w radianach
    # Adjust the angle sign based on the angular velocity
    if vel_ang < 0:
        gamma = -gamma

    return gamma


def cmd_vel_callback(self, msg):

    vel = np.array([msg.linear.x, msg.linear.y, msg.angular.z])

    steeringAngle1 = 0.0
    steeringAngle4 = 0.0
    if vel[2] != 0.0: 
        steeringAngle1 = calculate_angle(x1, y1, vel[0], vel[1], vel[2])
        steeringAngle4 = calculate_angle(x4, y4, vel[0], vel[1], vel[2])
    
    A = np.array([[np.sin(alpha1 + beta1 + steeringAngle1), -np.cos(alpha1 + beta1 + steeringAngle1), -l1*np.cos(beta1 + steeringAngle1)],
                  [np.sin(alpha2+beta2), -np.cos(alpha2+beta2), -l2*np.cos(beta2)],
                  [np.sin(alpha3+beta3), -np.cos(alpha3+beta3), -l3*np.cos(beta3)],
                  [np.sin(alpha4 + beta4 + steeringAngle4), -np.cos(alpha4 + beta4 + steeringAngle4), -l4*np.cos(beta4 + steeringAngle4)],
                  [np.cos(alpha1 + beta1 + steeringAngle1), np.sin(alpha1 + beta1 + steeringAngle1), l1*np.sin(beta1 + steeringAngle1)],
                  [np.cos(alpha2 + beta2), np.sin(alpha2 + beta2), l2*np.sin(beta2)],
                  [np.cos(alpha3 + beta3), np.sin(alpha3 + beta3), l3*np.sin(beta3)],
                  [np.cos(alpha4 + beta4 + steeringAngle4), np.sin(alpha4 + beta4 + steeringAngle4), l4*np.sin(beta4 + steeringAngle4)]])
    
    B = np.diag([0.033, 0.033, 0.033, 0.033, 0, 0, 0, 0])
    
    B_T = B.T                         # Transpose of B
    B_T_B_inv = np.linalg.pinv(B_T @ B)  # (B^T B)^{-1}
    B_T_A = B_T @ A                    # B^T A
    fi = B_T_B_inv @ B_T_A @ vel  # Final result for \dot{\phi}
    
    print(fi)

    jmsg = JointState()
    jmsg.velocity = fi.tolist()
    jmsg.position = [steeringAngle1, 0.0, 0.0, steeringAngle4]
    self.publisher_joint_states.publish(jmsg)
    

    
Robot.cmd_vel_callback = cmd_vel_callback
```

obliczanie pozycji!!!

```python
def theta_to_quaternion(theta):
    """
    Convert a 2D yaw (theta) angle into a quaternion.
    This is useful when you are working with 2D robot orientations.

    :param yaw: Yaw angle in radians
    :return: Quaternion corresponding to the yaw angle
    """
    # Compute the quaternion corresponding to the yaw angle
    q = Quaternion()
    q.z = np.sin(theta / 2.0)
    q.w = np.cos(theta / 2.0)
    return q

def joint_states_callback(self, msg):
    
    dt = 0.1
    
    x1 = 0
    y1 = -0.14
    alpha1 = (np.pi)/2
    beta1 = 0
    l1 = np.sqrt(x1*x1 + y1*y1)

    x2 = 0
    y2 = 0.14
    alpha2 = -(np.pi)/2
    beta2 = np.pi
    l2 = np.sqrt(x2*x2 + y2*y2)
    
    #-----------------------------
    
    fi = np.array(msg.velocity)
    
    J1 = np.array([[np.sin(alpha1+beta1), -np.cos(alpha1+beta1), -l1*np.cos(beta1)],
                   [np.sin(alpha2+beta2), -np.cos(alpha2+beta2), -l2*np.cos(beta2)]])
    
    J2 = np.diag([0.033, 0.033])
    
    R_inv = np.array([[np.cos(self.theta), -np.sin(self.theta), 0],
                      [np.sin(self.theta),  np.cos(self.theta), 0],
                      [0,                   0,                  1]])
    
    vel = np.linalg.multi_dot([R_inv, np.linalg.pinv(J1), J2, fi])
    
    self.x += vel[0]*dt
    self.y += vel[1]*dt
    self.theta += vel[2]*dt
    
    omsg = Odometry()
    omsg.header.frame_id = "odom"
    omsg.pose.pose.position.x = self.x
    omsg.pose.pose.position.y = self.y
    omsg.pose.pose.orientation = theta_to_quaternion(self.theta)
    self.publisher_odom.publish(omsg)
    
Robot.joint_states_callback = joint_states_callback
```

```python
# Initialize the ROS 2 Python client library
if not rclpy.ok():
    rclpy.init(args=None)
```

```python
def main(args=None):

    # Create an instance of the node
    node = Robot()


    # Spin the node to keep it active and handle callbacks
    while rclpy.ok():
        rclpy.spin_once(node)

    # Destroy the node when the process is finished
    #node.destroy_node()

    # Shutdown the ROS 2 Python client library
    # Spin the node or perform other ROS actions
    try:
        # rclpy.spin(node)  # Uncomment if you are using spinning
        pass
    finally:
        # Properly shut down ROS when done
        if rclpy.ok():
            rclpy.shutdown()

    
if __name__ == '__main__':
    main()
```
